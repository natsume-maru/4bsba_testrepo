# **Business Case**

Airbnb has transformed the way people travel, offering flexible short-term rentals across the globe. In highly competitive markets such as New York City and Singapore, thousands of listings compete for visibility and bookings.

However, hosts often struggle to price their properties effectively and to understand what drives occupancy. At the same time, investors and property managers want to know where to put their money and which property types yield the highest returns.

This creates a critical opportunity: leveraging data to identify high-performing neighborhoods, optimize pricing strategies, and reduce vacancy risk.

# **1. Defining the goal / problem:**

Airbnb hosts face several challenges:
- High competition: Too many similar listings drive price wars.
- Vacancy risk: Some neighborhoods are saturated while others are underserved.
- Price uncertainty: Hosts often price based on gut feeling, not data.
- Regulation pressure: Some areas limit or restrict short-term rentals, changing supply-demand dynamics.

How can Airbnb hosts and property investors maximize revenue and occupancy while minimizing risk in Singapore and New York City?
